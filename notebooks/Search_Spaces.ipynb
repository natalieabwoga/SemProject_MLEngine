## Google Maps Dijkstra's Algorithm Implementation

Suppose we want to find the shortest distance from B to F

- Each Key in the graph below is a `node` and each `node` has a dictionary of its neighbouring nodes and the distances to it
- For instance, node A's neighbours are B and C and the distance from A to B is 3 and the distance from A to C is 3

```python
graph = {
   "A": {"B": 3, "C": 3},
   "B": {"A": 3, "D": 3.5, "E": 2.8},
   "C": {"A": 3, "E": 2.8, "F": 3.5},
   "D": {"B": 3.5, "E": 3.1, "G": 10},
   "E": {"B": 2.8, "C": 2.8, "D": 3.1, "G": 7},
   "F": {"G": 2.5, "C": 3.5},
   "G": {"F": 2.5, "E": 7, "D": 10},
}
## This graph consists of nodes as the keys and a dictionary of the neighbours of the current node
```

## 1. Creating a Graph class

```python
from heapq import heapify, heappop, heappush
class Graph:
    def __init__(self, graph:dict={}):
        #The graph paramater here should be a dictionary
        self.graph = graph
    def add_edge(self, node1, node2, weight):
        if node1 not in self.graph:
            self.graph[node1]={}
            self.graph[node1][node2]=weight
    ##We create another class method, this is for the algorithm itself
    def shortest_distances(self, source: str):
        #Initializing the values of all nodes with infinity
        distances= {node: float("inf") for node in self.graph}
        # We set the source value to zero
        distances[source]=0
        pq=[(0,source)]
        heapify(pq)

        #We then create a set to hold visited nodes
        visited= set()

        #We start iterating over all unvisited nodes using a while loop
        while pq: #While the priority queue is not empty
            #We utilize tuple unpacking
            current_distance,current_node=heappop(pq)
            #We get the node with the minimal distance

            if current_node in visited:
                continue # Skip already visited nodes
            visited.add(current_node) # Else, add the node to visited nodes

            #We move on to visit its neighbours
            for neighbour, weight in self.graph[current_node].items():
                tentative_distance = current_distance + weight

                if tentative_distance < distances[neighbour]:
                    distances[neighbour]=tentative_distance
                    heappush(pq,(tentative_distance, neighbour))
            return distances
                    
                

            
```

## Uniform Cost Search

```python
highway_graph = {
    "Nairobi": [("Ruiru", 20), ("Naivasha", 90)],
    "Ruiru": [("Thika", 25)],
    "Thika": [("Nyeri", 110), ("Embu", 85)],
    "Embu": [("Meru", 75), ("Thika", 85)],
    "Meru": [("Nanyuki", 75)],
    "Nanyuki": [("Nyeri", 55)],
    "Nyeri": [("Nairobi", 115), ("Nyahururu", 80), ("Nanyuki", 55)],
    "Nyahururu": [("Nakuru", 105), ("Eldama Ravine", 70), ("Nyeri", 80)],
    "Eldama Ravine": [("Eldoret", 90), ("Nyahururu", 70)],
    "Nakuru": [("Naivasha", 65), ("Nyahururu", 105), ("Eldama Ravine", 70), ("Eldoret", 155), ("Molo", 50), ("Nyeri", 185)],
    "Naivasha": [("Nakuru", 65), ("Narok", 100), ("Nairobi", 90)],
    "Narok": [("Naivasha", 100), ("Bomet", 85)],
    "Bomet": [("Narok", 85), ("Kericho", 60), ("Kisii", 75)],
    "Kisii": [("Bomet", 75), ("Kisumu", 115)],
    "Kericho": [("Bomet", 60), ("Molo", 65), ("Kisumu", 100)],
    "Molo": [("Nakuru", 50), ("Kericho", 65)],
    "Eldoret": [("Webuye", 75), ("Kapsabet", 40), ("Eldama Ravine", 90), ("Nakuru", 155)],
    "Kapsabet": [("Eldoret", 40), ("Kisumu", 120)],
    "Webuye": [("Eldoret", 75), ("Kakamega", 40)],
    "Kakamega": [("Webuye", 40), ("Kisumu", 55)],
    "Kisumu": [("Kakamega", 55), ("Kapsabet", 120), ("Kericho", 100), ("Kisii", 115)]
}
```

```python
def uniform_cost_search(graph, start, goal):
    # At the frontier, 
    # The cost to to the frontier node is zero, the name of the node, and the path is the same
    frontier=[(0,start,[start])] 
    explored = set() # For explored nodes
    

    while frontier:
        cost, node, path = heappop(frontier)
        if node == goal:
            return cost, path
        if node in explored:
            continue
        explored.add(node)

        for neighbour, edge_cost in graph[node]:
            if neighbour not in explored:
                new_cost = cost + edge_cost
                heappush(frontier,(new_cost, neighbour, path + [neighbour]))
    return float("inf"),[] # In case we don't find any path
```

```python
total_cost, best_path = uniform_cost_search(highway_graph, "Nairobi","Kisumu")
print(best_path)
```
