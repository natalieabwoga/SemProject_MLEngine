```python
import pandas as pd
import numpy as np

columns = ["id", "label", "statement", "subject", "speaker",
           "speaker_job", "state_info", "party_affiliation",
           "barely_true_counts", "false_counts", "half_true_counts",
           "mostly_true_counts", "pants_on_fire_counts", "context"]

url = "https://raw.githubusercontent.com/thiagorainmaker77/liar_dataset/master/train.tsv"
dataset = pd.read_csv(url, sep='\t', names=columns)

dataset.head()
```

```python
rows, cols = dataset.shape

print("Rows:", rows)
print("Columns:", cols)
```

```python
print(dataset.dtypes)
```

```python
empty_per_col = dataset.isna().sum()
all_empty = empty_per_col.sum()

print(empty_per_col)
print("\nTotal empty cells:", all_empty)

if all_empty == 0:
    print("The dataset is complete.")
else:
    print(f"The dataset is incomplete. It is missing {all_empty} values.")
```

```python
set_head = dataset.head(15)
set_tail = dataset.tail(20)

df_sample = pd.concat([set_head, set_tail])

print("Shape:", df_sample.shape)
df_sample
```
